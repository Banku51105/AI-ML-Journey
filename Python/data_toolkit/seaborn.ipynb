```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Q1 — Distribution
scores = [
    45, 52, 67, 71, 73, 75, 78, 80, 81, 82,
    85, 87, 88, 90, 91, 92, 94, 96, 98, 100
]
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
sns.histplot(scores,bins=5,ax=ax[0])
sns.kdeplot(scores,ax=ax[1])
```

```python
# Q2 — Categorical Distribution
df = pd.DataFrame({
    "Department": [
        "ML", "ML", "ML",
        "Data", "Data", "Data",
        "Backend", "Backend", "Backend"
    ],
    "Salary": [
        70, 85, 95,
        60, 72, 80,
        75, 88, 92
    ]
})
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
sns.boxplot(x="Department",y="Salary",data=df,ax=ax[0])
sns.barplot(x="Department",y="Salary",data=df,ax=ax[1])
```

```python
# Q3 — Relationship + Category
df = pd.DataFrame({
    "Experience": [1, 2, 3, 4, 5, 6, 2, 4, 6],
    "Salary": [40, 45, 52, 60, 68, 80, 42, 65, 85],
    "Department": [
        "ML", "ML", "ML", "ML", "ML", "ML",
        "Data", "Data", "Data"
    ]
})
sns.scatterplot(x="Experience",y="Salary",data=df,hue="Department")
```
