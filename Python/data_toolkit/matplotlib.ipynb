```python
import matplotlib.pyplot as plt
```

```python
# Q1 — Basic Line Plot
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [120, 135, 128, 150, 165, 180]
plt.plot(months,revenue,marker="o")
plt.title("Monthly Revenue")
plt.xlabel("Months")
plt.ylabel("Revenue")
plt.grid()
```

```python
# Q2 — Bar Chart
departments = ["ML", "Data", "Backend", "Frontend"]
employees = [12, 18, 15, 10]
plt.bar(departments,employees)
plt.title("Employees by Department")
plt.xlabel("Departments")
plt.ylabel("Employees")
```

```python
# Q3 — Histogram
scores = [45, 52, 67, 71, 73, 75, 78, 80, 81, 82,
          85, 87, 88, 90, 91, 92, 94, 96, 98, 100]
plt.hist(scores,bins=5)
```

```python
# Q4 — Scatter
experience = [1, 2, 3, 4, 5, 6, 7]
salary = [40, 45, 52, 60, 68, 75, 85]
plt.scatter(experience,salary)
```

```python
# Q5 — Figure + Axes
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(months, revenue, marker="o")
axes[0].set_title("Monthly Revenue")
axes[1].bar(departments, employees)
axes[1].set_title("Employees by Department")
fig.tight_layout()
plt.show()
```

```python
# Q6 — Multiple Lines + Legend
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
product_a = [100, 120, 115, 140, 160, 175]
product_b = [90, 105, 125, 130, 145, 155]
product_c = [110, 115, 120, 125, 135, 150]
plt.plot(months,product_a,marker="o",label="Product A")
plt.plot(months,product_b,marker="o",label="Product B")
plt.plot(months,product_c,marker="o",label="Product C")
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Products Sold")
plt.legend()
plt.grid()
```

```python
# Q7 — Histogram Investigation
scores = [
    45, 52, 67, 71, 73, 75, 78, 80, 81, 82,
    85, 87, 88, 90, 91, 92, 94, 96, 98, 100
]
fig, ax = plt.subplots(1,2,figsize=(10, 4))
ax[0].hist(scores,bins=5)
ax[0].set_title("Scores with 5 Bins")
ax[1].hist(scores,bins=10)
ax[1].set_title("Scores with 10 Bins")
```

```python
# Q8 — Annotation
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [100, 120, 115, 150, 180, 170]
plt.plot(months,sales)
plt.annotate("Peak sales",xy=("May", 180),xytext=("Apr", 190),arrowprops={"arrowstyle": "->"})
```
