```python
# Real-World Problems
# Problem 1 — Dataset Preprocessing
data = np.array([
    [70, 80, 90],
    [60, 75, 85],
    [90, 95, 88],
    [45, 55, 60],
    [80, 85, 92]
])
student_averages = np.mean(data, axis=1)
print("Averages:", student_averages)
mask = student_averages > 80
top_students_data = data[mask]
print("Top students data:\n", top_students_data)
highest_per_subject = np.max(data, axis=0)
print("Highest in each subject (Math, Physics, Python):", highest_per_subject)
python_scores = data[:, 2]
highest_python_index = np.argmax(python_scores)
print("Index of highest Python score:", highest_python_index)
bonus_marks = np.array([5, 3, 2])
new_data = data + bonus_marks
new_averages = np.mean(new_data, axis=1)
print("New Data:\n", new_data)
print("New Averages:", new_averages)
```
