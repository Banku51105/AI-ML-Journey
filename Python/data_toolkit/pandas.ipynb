```python
# Q4 — Dtype Selection
df = pd.DataFrame({
    "Name": ["A", "B", "C"],
    "Age": [20, 21, 19],
    "Salary": [50000.0, 60000.0, 45000.0]
})
print(df)
df.select_dtypes(include="number")
df.select_dtypes(include="str")
```

```python
# Q5 — Boolean Indexing
df = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E"],
    "Age": [17, 25, 31, 16, 28],
    "Score": [80, 65, 92, 74, 88]
})
print(df[df["Age"] > 18])
print(df[df["Score"] > 80])
print(df[(df["Age"] > 18) & (df["Score"] > 80)])
print(df[(df["Age"] > 18) & (df["Score"] > df["Score"].mean())].Name)
```

```python
# Q6 — Missing Values
df = pd.DataFrame({
    "A": [10, 20, None, 40],
    "B": [1, None, 3, 4],
    "C": [None, None, None, 5]
})
print(df)
print(df.isnull().sum())
print(df.dropna())
print(df.dropna(axis=1))
print(df["A"].fillna(df["A"].mean()))
print(df["B"].fillna(0))
```

```python
# Q7 — Duplicates
df = pd.DataFrame({
    "Name": ["A", "B", "A", "C", "B"],
    "Score": [80, 90, 80, 70, 90]
})
print(df[df.duplicated()])
print(df.duplicated().sum())
print(df.drop_duplicates())
```

```python
# Q8 — GroupBy
df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "Sales"],
    "Employee": ["A", "B", "C", "D", "E"],
    "Salary": [50000, 60000, 45000, 55000, 70000]
})
print(df.groupby("Department")["Salary"].mean())
print(df.groupby("Department")["Salary"].max())
print(df.groupby("Department")["Employee"].size())
```

```python
# Q9 — Handling Missing Values and Aggregating Data
df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "Employee": ["A", "B", "C", "D", "E", "F"],
    "Salary": [50000, 60000, 45000, None, 70000, 65000]
})
df["Salary"].isnull()
df.fillna({"Salary": df["Salary"].median()}, inplace= True)
df.groupby("Department")["Salary"].mean()
df.groupby("Department")["Salary"].mean().idxmax()
df
```

```python
# Q10 — String Operations
df = pd.DataFrame({
    "text": [
        "Hello Data Science",
        "I love Machine Learning",
        "Pandas is useful"
    ]
})
df["Lower"] = df["text"].str.lower()
df["Length"] = df["text"].str.len()
df["Startswith"] = df["text"].str.startswith("I")
df["Words"] = df["text"].apply(lambda x : len(x.split()))
df
```

```python
# Q11 — Datetime
df = pd.DataFrame({
    "date": ["2026-01-15", "2026-02-20", "2026-03-10"]
})
df["Date"] = pd.to_datetime(df["date"])
print(df["Date"].dt.year)
print(df["Date"].dt.month)
print(df["Date"].dt.day)
print(df["Date"].dt.day_name())
df["7 Days Later"] = df["Date"] + pd.Timedelta(days=7)
df
```

```python
# Master Practice Set
# Q1 Inspect, Selection, Missing Values, Group Analysis and Feature Creation
df = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E", "F"],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "IT"],
    "Age": [22, 25, 21, 30, np.nan, 24],
    "Salary": [50000, 45000, 55000, 70000, 48000, np.nan],
    "Experience": [1, 3, 1, 6, 4, 2]
})
print(df)

print(df.shape)
print(df.dtypes)
print(df.isnull().sum())
print(df.describe())

print(df[["Name","Salary"]])
print(df[df["Age"]>23])
print(df[["Name","Salary"]][df["Experience"] >= 3])

print(df.isnull())
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print(df.groupby("Department")["Salary"].mean())
print(df.groupby("Department")["Experience"].max())
print(df.groupby("Department")["Name"].size())
print(df.groupby("Department")["Salary"].mean().idxmax())

df["Annual_Salary_per_Experience"] = df["Salary"]/df["Experience"]
df
max_index = df['Annual_Salary_per_Experience'].idxmax()
top_employee = df.loc[max_index, 'Name']
print(top_employee)
```

```python
# Q2
employees = pd.DataFrame({
    "Employee": ["A", "B", "C", "D", "E", "F", "G"],
    "Team": ["ML", "ML", "Data", "Data", "ML", "Backend", "Backend"],
    "Experience": [2, 5, 3, np.nan, 4, 6, 1],
    "Salary": [50000, 90000, 65000, 70000, np.nan, 85000, 45000],
    "Projects": [2, 7, 4, 5, 6, 8, 1]
})
print(employees)
employees["Experience"] = employees["Experience"].fillna(employees["Experience"].median())
employees["Salary"] = employees["Salary"].fillna(employees["Salary"].median())
print(employees[(employees["Experience"] > 3) & (employees["Projects"] >= 5)][["Employee", "Team", "Salary",]])
average_salary_of_each_team = employees.groupby("Team")["Salary"].mean()
print(average_salary_of_each_team)
team_with_the_highest_average_salary = average_salary_of_each_team.idxmax()
print(team_with_the_highest_average_salary)
employees["Projects_per_Year"] = employees["Projects"]/employees["Experience"]
employees.loc[employees["Projects_per_Year"].idxmax(),"Employee"]

bonus = pd.DataFrame({
    "Employee": ["A", "B", "C", "E", "F"],
    "Bonus": [5000, 10000, 7000, 8000, 12000]
})
employees = pd.merge(employees,bonus,how="left")
print(employees["Employee"][employees["Bonus"].isnull()])
employees["Bonus"] = employees["Bonus"].fillna(0)
employees["Total_Compensation"] = employees["Salary"] + employees["Bonus"] 
employees.loc[employees[employees["Team"] == "ML"]["Total_Compensation"].idxmax(),"Employee"]
```
